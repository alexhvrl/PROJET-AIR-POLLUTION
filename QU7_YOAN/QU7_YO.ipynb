```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import shapiro, anderson, normaltest, probplot, kstest, norm

plt.rcParams.update({
    'axes.titlesize': 18,
    'axes.labelsize': 16,
    'xtick.labelsize': 14,
    'ytick.labelsize': 14
})

# 1. Load and Clean Data
file_path = 'DATA.csv'  # Replace with your file
df = pd.read_csv(file_path, sep=';', decimal=',')
df['TIME'] = pd.to_datetime(df['TIME'], dayfirst=True)
df['month'] = df['TIME'].dt.month
july_df = df[df['month'] == 7]

# 2. Define pollutant columns for each site
pollutants = {
    'CHA': ['NO2CHA', 'O3CHA', 'PM10CHA'],
    'MAG': ['NO2MAG', 'O3MAG', 'SO2MAG', 'PM10MAG', 'PM2.5MAG']
}

# 3. Process: hourly and daily for each
def analyze_distribution(data, label, is_daily=False):
    data = pd.to_numeric(data, errors='coerce')  # Convert to numeric, set bad entries to NaN
    data = data.dropna()  # Drop NaNs

    data = data[data > 0]  # Only positive values (log can't handle zero or negatives)
    if len(data) < 10:
        print(f"[{label}] Not enough data for normality test.")
        return

    #log_data = data 
    log_data = data
    #log_data = np.log(data)

    # Normality Tests
    # Compute statistics
    mean_val = np.mean(log_data)
    std_val = np.std(log_data)
    shapiro_stat, shapiro_p = shapiro(log_data)
    anderson_result = anderson(log_data)
    ks_stat, ks_p = kstest(log_data, 'norm', args=(mean_val, std_val))


    # Print results
    print(f"\n{'DAILY' if is_daily else 'HOURLY'} - {label}")
    print(f"Mean: {mean_val:.4f}")
    print(f"Standard Deviation: {std_val:.4f}")
    print(f"Shapiro-Wilk stat: {shapiro_stat:.4f}")
    print(f"Shapiro-Wilk p-value: {shapiro_p:.4f}")
    print(f"Anderson-Darling stat: {anderson_result.statistic:.4f}")
    print("Anderson critical values:", anderson_result.critical_values)
    print(f"Kolmogorov-Smirnov stat: {ks_stat:.4f}")
    print(f"Kolmogorov-Smirnov p-value: {ks_p:.4f}")

    # Plots
    fig, axs = plt.subplots(1, 3, figsize=(16, 4))
    #sns.histplot(data, kde=True, ax=axs[0])
    #axs[0].set_title(f'Histogram: {label}')

    sns.histplot(log_data, kde=True, ax=axs[0])
    axs[0].set_xlabel(f"Log({label.split()[0]})")
    axs[0].set_title(f'Histogram: {label}', size=18)
    axs[0].text(0.95, 0.95,
                f"μ = {mean_val:.2f}\nσ = {std_val:.2f}\nW = {shapiro_stat:.4f}\np = {shapiro_p:.4f}",
                transform=axs[0].transAxes,
                verticalalignment='top',
                horizontalalignment='right',
                bbox=dict(facecolor='white', edgecolor='black', boxstyle='round,pad=0.4'), size=12)


    #sns.ecdfplot(data, ax=axs[1])
    #axs[1].set_title(f'ECDF: {label}')

    sns.ecdfplot(log_data, ax=axs[1])
    axs[1].set_xlabel(f"Log({label.split()[0]})")
    axs[1].set_title(f'ECDF: {label}', size=18)

    probplot(log_data, dist="norm", plot=axs[2])
    axs[2].set_title(f'QQ Plot: {label}', size=18)
    plt.tight_layout()
    plt.show()

for site, cols in pollutants.items():
    for col in cols:
        # Convert column to numeric before analysis
        july_df[col] = pd.to_numeric(july_df[col], errors='coerce')

        # Hourly
        analyze_distribution(july_df[col], f"{col} (Hourly)")

        # Daily
        daily_data = july_df.set_index('TIME')[col].resample('D').mean()
        analyze_distribution(daily_data, f"{col} (Daily)", is_daily=True)
```
